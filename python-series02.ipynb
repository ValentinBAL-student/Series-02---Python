This is the notebook to achieve the assignment 02 of the python series, the goal is to plot a sinusoid signal from math concept learned in class to code it in python.


To generate a sine wave, we need this formule.
y(t)=A*sin(ωt+φ)

Where A mean the amplitude, unit depedn on the physical 
sin is the function 
ω mean the angular frequency
t is time, here in second
and φ is the initial angle at time = 0

```python
#A.import the necessary libraries to complete the assignment
import numpy as np
import matplotlib.pyplot as plt

#B.Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0,1,1000)
#C.Create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = np.sin(2*np.pi*t)
#C.a Create  a second numpy array called s2 that contains the values of a cosine function of t
s2 = np.cos(2*np.pi*t)
# Plot s as a sine function of t
plt.plot(t, s)
# Plot s2 as a cosine function of t
plt.plot(t, s2)
#Setting of the graphical parameters of the plot
# Add a title to the plot
plt.title("Sine and Cosine Waves")
# Add a label to the x axis
plt.xlabel("Time")
# Add a label to the y axis
plt.ylabel("Amplitude")
#Add a legend to the plot
plt.legend(["Sine", "Cosine"])
# Add a grid to the plot
plt.grid(True)
# Save the plot as a png file
plt.savefig("cosine_and_sine_wave.png")
plt.show()
```
