This is the notebook to achieve the assignment 02 of the python series, the goal is to plot a sinusoid signal from math concept learned in class to code it in python.


To generate a sine wave, we need this formule.
y(t)=A*sin(ωt+φ)

Where A mean the amplitude, unit depedn on the physical 
sin is the function 
ω mean the angular frequency
t is time, here in second
and φ is the initial angle at time = 0

```python
#import the necessary libraries to complete the assignment
import numpy as np
import matplotlib.pyplot as plt

#Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0,1,1000)
#Create a function called sinusoid that:
#takes as input all parameters that you need to define the sinusoid
def sinusoid(t, amplitude, frequency, phase):
    return amplitude * np.sin(2 * np.pi * frequency * t + phase)
#Use the function sinusoid to create two numpy arrays s1 and s2 with the two different frequencies
s1 = sinusoid(t, 1, 1.5, 0)
s2 = sinusoid(t, 1, 3, 0)
# Plot s1 and s2 as a function of t
plt.plot(t, s1, label="S1")
plt.plot(t, s2, label="S2")
#Setting of the graphical parameters of the plot
# Add a title to the plot
plt.title("Sine and Cosine Waves")
# Add a label to the x axis
plt.xlabel("Time")
# Add a label to the y axis
plt.ylabel("Amplitude")
#Add a legend to the plot
plt.legend(["Sin1 = 1.5 Hz", "Sin2 = 3 Hz"])
# Add a grid to the plot
plt.grid(True)
# Save the plot as a png file
plt.savefig("sine_waves.png")
plt.show()
```
